## Singleton

(The Constant Function)

---

Let $X$ be the set of possible inputs (ignored at runtime) and let $\mathcal{O}$ be the set of all instances of the class. The Singleton defines:

$$f : X \rightarrow \{c\}, \qquad c \in \mathcal{O}$$

a constant function whose codomain is a singleton set — every call returns the same object $c$.

$$f(1) = c, \quad f(2) = c, \quad f(99) = c$$

All calls return the same object by identity:

$$f(x_1) \equiv_{\mathrm{id}} f(x_2) \quad \forall\, x_1, x_2 \in X$$

**Uniqueness condition** — the image of $f$ contains exactly one element:

$$|\mathrm{im}(f)| = 1, \qquad \mathrm{im}(f) = \{c\}$$

Equivalently: $\exists!\, c \in \mathcal{O}$ such that $f(x) = c$ for all $x \in X$.

**Mutation property** — since $f(x_1) \equiv_{\mathrm{id}} f(x_2)$, any mutation at time $t_1$ is immediately visible through any other reference:

$$\mu(f(x_1),\, t_1) \Rightarrow \mathrm{state}(f(x_2),\, t_2) \text{ reflects } \mu \quad \forall\, x_2,\ t_2 > t_1$$

This is both the *power* and the *danger* of Singleton: shared state is global.

**Contrast with Factory Method:**

$$\text{Singleton}: \quad |\mathrm{im}(f)| = 1 \quad \text{(same object every call)}$$

$$\text{Factory Method}: \quad f(k)_1 \not\equiv_{\mathrm{id}} f(k)_2 \quad \text{(fresh object every call)}$$

### How to decide if you need a Singleton?
Ask yourself these three questions:

- "Should only one of these exist?" → Yes
- "Do multiple parts of my app need it?" → Yes
- "Does having two break something?" → Yes

If all three are yes: Singleton. If any one is no: reconsider.


### Exercise 1 — Basic Singleton
---
 
**Scenario:** Imagine your app connects to a database. Every time a new connection is created, it costs time and memory. A Singleton ensures only *one* connection is ever made.
 
**Your task:** Create a `DatabaseConnection` class. No matter how many times someone calls it, it should return the exact same object.
 
```python
db1 = DatabaseConnection()
db2 = DatabaseConnection()
 
print(db1 is db2)  # Should print: True
```

````python
class DatabaseConnection:
    _instance = None

    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
        return cls._instance


db1 = DatabaseConnection()
db2 = DatabaseConnection()
print(id(db1))
print(id(db2))
print(db1 is db2) # Should print: True
````

````python
# Suppose  we don't use the __new__ method, what will happen?

class DatabaseConnection2:
    pass

db3 = DatabaseConnection2()
db4 = DatabaseConnection2()

print(id(db3))
print(id(db4))
print(db3 is db4) # Should print: False
````

### Exercise 2 — Thread-Safe Singleton
---
 
**Scenario:** What if two parts of your app try to create the Singleton at the *exact same time*? You could end up with two instances. This exercise fixes that.
 
**Your task:** Extend Exercise 1 so it's safe when multiple threads run simultaneously.


````python
import threading

class DatabaseConnection3:
    _instance = None
    _lock = threading.Lock()

    def __new__(cls):
        if cls._instance is None:
            with cls._lock:
                if cls._instance is None:
                    cls._instance = super().__new__(cls)
        return cls._instance


db5 = DatabaseConnection3()
db6 = DatabaseConnection3()
print(id(db5))
print(id(db6))
print(db5 is db6)     # Should print: True
````
